## Саммари

**Что разобрали:**
- **(а) Хотя бы одна решка:** $1 - P(\text{ОООО}) = 1 - 0.5^4 = 0.9375$.
- **(б) Ровно два орла:** 6 подходящих цепочек, каждая с вероятностью $0.5^4 = \frac1{16}$, итого $6 \cdot 0.0625 = 0.375$.
- **Почему так удобно:** в (а) неподходящая цепочка одна (`ОООО`), а подходящих 15, поэтому проще посчитать одну и вычесть из 1. В (б) $6 \cdot 0.0625$ — короткая запись суммы шести одинаковых вероятностей.
- **ML:** «хотя бы один = 1 − ни одного». Ошибка 1% на примере даёт $1 - 0.99^{100} \approx 63\%$ шанс хотя бы одной ошибки на 100 примерах.
- **Повторение:** $\log_2 16 = 4$ (16 цепочек $= 2^4$, 4 броска); $\ln(0.5^4) = 4\ln 0.5$.

### Теория по трудным местам

**Исход эксперимента — вся цепочка.**
- Эксперимент — 4 броска. Его исход — цепочка из 4 букв, например `ОРОР`. Всего цепочек $2^4 = 16$.
- Порядок важен: `ООРР` и `ОРОР` — разные исходы, хотя орлов в них поровну.

**$P(A)$ и $P(A \mid B)$.**
- $P(A)$ — просто вероятность $A$. $P(A \mid B)$ — вероятность $A$, *если известно*, что произошло $B$.
- Для одного броска пишем $P(\text{О}) = 0.5$, без условия.

**«Хотя бы одна» — это $\ge 1$, а не «ровно одна».**

**Вероятность конкретной цепочки.**
- Броски независимы, поэтому вероятности бросков внутри цепочки **перемножаются**: $P(\text{ОРОР}) = 0.5 \cdot 0.5 \cdot 0.5 \cdot 0.5 = \frac1{16}$.
- У любой цепочки из 4 бросков такая же вероятность, $0.5^4$, а не $0.5$.

**Независимость и взаимоисключаемость — разные вещи.**
- **Независимые** события не влияют друг на друга (разные броски). Для «$A$ и $B$» вероятности умножаем.
- **Взаимоисключающие** события не могут произойти в **одном** прогоне (две разные цепочки). Для «$A$ или $B$» вероятности складываем.
- В следующем прогоне та же цепочка может выпасть снова, это взаимоисключаемости не противоречит.

**Для расчёта вероятности число повторов не нужно.**
- Вероятность — свойство одного прогона. «Доля при большом числе повторов» — способ её *проверить*: в 10 000 прогонах `ОООО` выпадет примерно в 6.25% случаев.

**Сколько цепочек с ровно двумя О.**
- Цепочку задают позиции двух О: {1,2}, {1,3}, {1,4}, {2,3}, {2,4}, {3,4}. Получается 6.
- Это число обозначают $\binom{4}{2}$, «C из 4 по 2». Формулу разберём, когда понадобится.
