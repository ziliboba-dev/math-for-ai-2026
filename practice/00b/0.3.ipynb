## Разведка: значения $f(x)$ при $x = 1 \ldots 5$

Впадина видна по таблице — наименьшее значение в середине.

````python
for x in range(1, 6):
    fx = (x-3)**2 + 1
    print(fx)
````

## Способ 1. Через производную

Раскрыть скобку → продифференцировать почленно → приравнять $f'(x)$ к нулю.

```text
(x-3)**2 + 1
(x**2 - 6*x + 9) + 1
x**2 - 6*x + 10 

(x**2)' = 2*x
(-6*x)' = -6
(1)' = 0

2x-6
```

## Побочный вопрос: $f'(x) = 0$ — всегда минимум?

Проверка на $y = -x^2$: производная $-2x$.

```text
-2x=0
x=0
```

## Способ 2. Градиентный спуск

Формула шага (`n` — это $\eta$, learning rate):

```text
Gradient descent

x <- x - nf'(x)
```

Функции наклона и шага, стартовые значения:

````python
grad = lambda x: 2*x-6
step = lambda x: x - n * grad(x)

x = 0
n = 0.1
````

50 шагов спуска из $x = 0$:

````python
for i in range(50):
    x = step(x)
    print(x)
````

## Саммари

**Что разобрали:**
- **Минимум** — самая низкая точка графика, у неё две координаты: где ($x$) и какая высота ($f(x)$). Ответ: $(3, 1)$.
- **Решение без производной:** $(x-3)^2 \ge 0$ и равно нулю при $x = 3$, значит $f \ge 1$, и значение $1$ достигается.
- **Способ через производную:** во впадине наклон меняет знак с минуса на плюс, а в точке перехода равен $0$. Решаем $f'(x) = 0$: $2x - 6 = 0 \Rightarrow x = 3$.
- **$f' = 0$ находит только кандидатов.** У $-x^2$ в нуле максимум, у $x^3$ ровное место. Какой случай перед нами, проверяют отдельно.
- **Градиентный спуск:** $x \leftarrow x - \eta f'(x)$. Шаг идёт против наклона, поэтому с любой стороны движение к впадине. У дна наклон стремится к $0$, и шаги затухают.
- **Сверх задачи:** расстояние до минимума умножается на $1 - 2\eta = 0.8$ за шаг.

### Теория по трудным местам

**Минимум и область определения — разные вещи.**
- *Минимум* — наименьшее значение $f$ и точка $x$, где оно достигается.
- *Область определения* — все $x$, при которых формула имеет смысл. Например, $\ln x$ определён только при $x > 0$.

**Производная суммы и множителя.**
- $(u + v)' = u' + v'$, $(c \cdot u)' = c \cdot u'$.
- Знак входит в коэффициент: $(-6x)' = -6$.
- Проверка: в найденном минимуме должно выполняться $f'(x) = 0$.

**Порядок действий: степень раньше унарного минуса.**
- $-x^2 = -(x^2)$, это не то же самое, что $(-x)^2 = x^2$.
- Так же в Python: `-1**2 == -1`.
- $c \cdot f(x)$ при $c = -1$ отражает график относительно оси $x$, поэтому $-x^2$ — перевёрнутая парабола. Гипербола — это $1/x$.

**Почему $f'(x) = 0$ во впадине.**
- Производная — это наклон касательной.
- Слева от впадины функция убывает ($f' < 0$), справа растёт ($f' > 0$). В точке перехода касательная горизонтальна, $f' = 0$.
- Поэтому минимум ищут как решение уравнения $f'(x) = 0$.

**Цепное правило (функция от функции).**
- $(f(g(x)))' = f'(g(x)) \cdot g'(x)$: производная внешней функции, где внутренняя остаётся нетронутой, умноженная на производную внутренней.
- Как распознать вложенность: если в $(\ldots)^2$ вместо голого $x$ стоит выражение, это функция от функции.
- Порядок вычисления $(x-3)^2$: сначала $g = x - 3$, потом квадрат.

**Определение производной; $h$ и $\eta$.**
- $f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$, где $h$ — пробный сдвиг для измерения наклона, $h \to 0$.
- $\eta$ (learning rate) — настоящий шаг спуска, конечное число. Сдвиг $x$ за шаг равен $\eta \cdot f'(x)$, а не $\eta$.

**Почему спуск идёт к впадине.**
- Алгоритм не знает, где минимум, он видит только наклон в текущей точке.
- Минус в $x - \eta f'(x)$ означает шаг против наклона: при $f' < 0$ вправо, при $f' > 0$ влево.

**Присваивание и состояние в коде.**
- $\leftarrow$ в формуле означает присваивание: каждый шаг сначала считает наклон в **текущем** $x$, потом записывает в $x$ новое значение.
- В ноутбуке переменные живут между запусками ячеек: повторный запуск цикла продолжит с последнего $x$.

**Вынесение общего множителя.**
- $a - k \cdot a = a(1 - k)$.
- Для расстояния до минимума $d = x - 3$: $d_{\text{new}} = d - 2\eta d = d(1 - 2\eta)$.
