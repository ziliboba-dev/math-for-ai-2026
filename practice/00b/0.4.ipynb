# Модуль 00½ · Задача 0.4. Средний лосс

Вероятности правильного класса: $0.9$, $0.6$, $0.99$. Найти $-\frac{1}{3}\sum \ln p$.

`modules/00b_basics_short.md`

```python
import numpy as np
```

```python
p1 = 0.9
p2 = 0.6
p3 = 0.99
```

## Логарифмы вероятностей

`np.log` — натуральный логарифм. Все три значения отрицательные, потому что $p < 1$.

```python
res1 = np.log(p1)
res2 = np.log(p2)
res3 = np.log(p3)

print(res1, res2, res3)
```

```python
np.log(0.9*0.6*0.99)
```

## Средний лосс

Минус переворачивает знак, деление на 3 — среднее по трём примерам.

```python
-(res1 + res2 + res3)/3
```

## Саммари

**Что разобрали:**
- **Ответ:** $-\frac13(\ln 0.9 + \ln 0.6 + \ln 0.99) = (0.105 + 0.511 + 0.010)/3 \approx 0.209$.
- **Смысл формулы:** средний штраф за то, что модель недодала вероятность правильному ответу. $0$ — идеально, чем больше, тем хуже.
- **Главный вклад** даёт пример с $p = 0.6$: $0.51$ из $0.63$. Модель в нём меньше всего уверена.
- **Это кросс-энтропия** (CE-лосс), основной лосс классификаторов и LLM. Обучение уменьшает это число, то есть поднимает $p$ правильного ответа.
- **Повторение:** $\ln 0.9 + \ln 0.6 + \ln 0.99 = \ln(0.9 \cdot 0.6 \cdot 0.99) = \ln 0.5346 \approx -0.626$, сверено численно.

### Теория по трудным местам

**Что такое $p$.**
- $p$ — вероятность, которую модель дала **правильному ответу в конкретном примере**.
- Три числа — это три примера, а не три класса. Правильный класс в разных примерах может быть разным.

**Знак $\ln x$ определяется положением $x$ относительно 1.**
- $\ln x$ — степень, в которую надо возвести $e$, чтобы получить $x$. $e^0 = 1$.
- $0 < x < 1 \Rightarrow \ln x < 0$; $x = 1 \Rightarrow \ln x = 0$; $x > 1 \Rightarrow \ln x > 0$.
- Вероятность $\le 1$, поэтому $\ln p \le 0$ сам по себе, без всяких множителей впереди.

**Зачем минус.**
- Без минуса: $\ln 0.99 \approx -0.01 > \ln 0.1 \approx -2.3$, то есть хорошее предсказание даёт **большее** число. Для штрафа это наоборот.
- С минусом: $0.01$ против $2.3$. Лосс $\ge 0$, растёт, когда предсказание хуже, и равен $0$ при $p = 1$.
- Поэтому «минимизировать лосс» = «поднимать $p$ правильного ответа».

**$\frac1N\sum$ — среднее.**
- Сумма по примерам, делённая на их число. Множитель $\frac1N$ можно применить к каждому слагаемому, результат тот же.

**Логарифм произведения.**
- $\ln(abc) = \ln a + \ln b + \ln c$: логарифм превращает произведение в сумму (`modules/01_notation.md`).
- Перемножать надо сами числа, а не их логарифмы. Произведение отрицательных логарифмов может быть $< 0$, а $\ln$ от него не определён, поэтому получается `nan`.

**Связь с полной формулой кросс-энтропии** (модуль 08, сверх задачи).
- Для одного примера: $H(P, Q) = -\sum_x p(x)\log q(x)$, где $P$ — правильный ответ (one-hot), $Q$ — выход модели.
- В one-hot все $p(x)$, кроме верного класса, равны $0$, и остаётся $-\log q(y_{\text{true}})$.
- Буквы в модуле 08 переставлены: там $p$ — правильный ответ, $q$ — модель. В 0.4 $p$ — это модель.
