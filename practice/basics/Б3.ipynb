- **Б3.1.** Найдите область определения $f(x) = \ln(1 - x^2)$.
- **Б3.2.** Покажите, что логит обратен к сигмоиде.
- **Б3.3.** Где «ступенька» $\sigma(4x - 8)$ проходит через 0.5 и насколько она крутая по сравнению с $\sigma(x)$?

```python
import numpy as np

sigma = lambda z: 1 / (1 + np.exp(-z))
logit = lambda p: np.log(p / (1 - p))
```

```python
# Б3.1: внутри (-1, 1) — число, на границе -inf, снаружи nan
for x in [-2, -1, -0.5, 0, 0.5, 1, 2]:
    print(x, np.log(1 - x**2))
```

```python
# Б3.2: logit(sigma(z)) возвращает z
for z in [-3, 0, 2.5]:
    print(z, logit(sigma(z)))
```

```python
# Б3.3: центр ступеньки и наклон в центре
h = 1e-4
f = lambda x: sigma(4 * x - 8)
print(f(2))                                               # 0.5
print((f(2 + h) - f(2)) / h, (sigma(h) - sigma(0)) / h)   # ~1.0 против ~0.25
```

## Саммари

**Что разобрали:**
- **Б3.1:** $\ln$ определён только для положительного аргумента: $1 - x^2 > 0 \iff x^2 < 1 \iff -1 < x < 1$.
- **Б3.2:** $\mathrm{logit}(\sigma(z)) = z$. Подставили $p = \frac{1}{1+e^{-z}}$ в $\ln\frac{p}{1-p}$:

$$\frac{p}{1-p} = \frac{\frac{1}{1+e^{-z}}}{\frac{e^{-z}}{1+e^{-z}}} = \frac{1}{e^{-z}} = e^{z}, \qquad \ln e^{z} = z$$

- **Б3.3:** $\sigma(4x-8) = 0.5$, когда аргумент равен 0: $4x - 8 = 0$, $x = 2$. Множитель 4 при $x$ сжимает график по горизонтали в 4 раза, поэтому ступенька в 4 раза круче: наклон в центре $4 \cdot 0.25 = 1$ против $0.25$ у $\sigma(x)$.
- **Повторение (interleaving):** вероятность двух шестёрок подряд $\frac16 \cdot \frac16 = \frac{1}{36}$.
- **ML:** нейрон $\sigma(wx + b)$ — ступенька. $w$ задаёт её крутизну, точка $x = -b/w$ — положение. Логит — это то, что модель выдаёт до сигмоиды.

### Теория по трудным местам

**Неравенство с квадратом:**
- $x^2 < a \iff \lvert x\rvert < \sqrt a \iff -\sqrt a < x < \sqrt a$ (отрезок вокруг нуля);
- $x^2 > a \iff x < -\sqrt a$ или $x > \sqrt a$ (два луча).

Проверка числом: $x = -2$ даёт $x^2 = 4 > 1$, значит, условие $x < 1$ не годится.

**Обратная функция** отменяет исходную: $g(f(z)) = z$, то есть на выходе исходный аргумент, а не 0. Пары: $\ln$ и $e^x$, $\mathrm{logit}$ и $\sigma$.

**Логарифм экспоненты:** $\ln e^{z} = z$. Не путать с производной: $(e^z)' = e^z$.

**Логарифм отрицательного числа** в вещественных числах не определён, поэтому `np.log(-1)` даёт `nan`. В комплексных числах $\ln(-1) = i\pi$, но в курсе это не нужно.

**Преобразования графика:**
- $f(kx)$ — сжатие по **горизонтали** в $k$ раз: всё происходит в $k$ раз быстрее, наклон умножается на $k$;
- $c \cdot f(x)$ — растяжение по **вертикали**: меняется высота, а не скорость.

Множитель внутри аргумента действует по горизонтали, снаружи — по вертикали.

**«Ступенька»** — график, который слева почти 0, справа почти 1, а между ними плавный переход. Это сигмоида; её центр там, где аргумент равен 0.

**Независимые события → умножение:** $P(A \text{ и } B) = P(A)\,P(B)$. Вероятность выпадения шестёрки на кубике $\frac16$, а не $\frac12$: у кубика 6 равновероятных граней.
