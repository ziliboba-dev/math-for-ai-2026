## Саммари

**Что разобрали:**
- **Разминка:** $A = \{1,2,3\}$, $B = \{2,3,4\}$ → $A \cap B = \{2,3\}$, $A \cup B = \{1,2,3,4\}$.
- **Б12.1:** train ∩ test = id 750–800, всего $800 - 750 + 1 = 51$ пример. Это утечка данных: модель видела часть теста при обучении, и оценка качества **завышена**. Должно быть train ∩ test = ∅.
- **Б12.2:** $|A \cup B| = 40 + 60 + 20 = 100 + 80 - 60 = 120$, $IoU = \frac{60}{120} = 0.5$. Крайние случаи: рамки совпали → 1, не пересекаются → 0.
- **Б12.3:** «Для любого $\varepsilon > 0$ существует номер $N$ такой, что для всех $n > N$ расстояние $|a_n - L|$ меньше $\varepsilon$» — это определение предела $a_n \to L$. Пример: $a_n = \frac1n$, $L = 0$: при $\varepsilon = 0.1$ подходит $N = 10$, при $\varepsilon = 0.01$ — $N = 100$.
- **Повторение (interleaving):** $x^2 < 9 \iff -3 < x < 3$.
- **ML:** train ∩ test = ∅ — проверка на утечку; IoU — метрика детекции и сегментации; «loss сходится» — это предел.

### Теория по трудным местам

**Пересечение и объединение:**
- $x \in A \cap B \iff x \in A$ **и** $x \in B$;
- $x \in A \cup B \iff x \in A$ **или** $x \in B$.

«Или» в математике нестрогое: годится и одно, и оба сразу. Поэтому общие элементы тоже входят в объединение, но по одному разу.

**Мощность объединения:** $|A \cup B| = |A| + |B| - |A \cap B|$. Пересечение вычитаем, потому что оно уже посчитано и в $|A|$, и в $|B|$. Тот же результат даёт разбиение на кусочки: «только A» + «оба» + «только B».

**Моделирование в коде:** множества нужно строить так, чтобы совпадали все три числа из условия: $|A|$, $|B|$ и $|A \cap B|$. Если `B = np.arange(0, 80)` лежит внутри `A = np.arange(0, 100)`, общих элементов получится 80, а не 60. `np.arange(a, b)` и `range(a, b)` не включают $b$, в них $b - a$ элементов.

**Количество целых от $a$ до $b$ включительно:** $b - a + 1$. Проверка на маленьком примере: от 1 до 3 — это 3 числа.

**Утечка данных завышает качество:** модель не угадывает ответ на тестовом примере, а помнит его с обучения.

**Чтение кванторов:**
- «:» — «такой, что»;
- «$\forall n > N$» — «для всех $n$, больших $N$». Это начало нового условия, а не свойство $N$;
- $|a - b|$ — расстояние между числами.

**Предел:** $a_n \to L$, если начиная с некоторого номера члены последовательности остаются ближе к $L$, чем любая заданная точность $\varepsilon$. Число $L$ называется пределом, запись: $\lim_{n\to\infty} a_n = L$.

**Неравенство с квадратом:** $x^2 < a \iff -\sqrt a < x < \sqrt a$. У ответа две границы: отрицательные $x$ с большим модулем тоже дают большой квадрат. Проверка: $x = -4$ удовлетворяет $x < 3$, но $(-4)^2 = 16 > 9$.

**Решение неравенства и область определения:** решение неравенства — это все $x$, при которых неравенство истинно. Область определения — это все $x$, при которых функция имеет смысл. Чтобы найти область определения, часто решают неравенство: например, для $\ln(1-x^2)$ решают $1 - x^2 > 0$.

**Хвосты:** нет. Ошибка $x^2 < a \Rightarrow x < \sqrt a$ повторяется второй раз (после Б3.1), поэтому стоит вернуться к ней позже.
