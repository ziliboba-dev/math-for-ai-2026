# Модуль 01 · Задача 1.1. Производная сигмоиды

Докажите, что $\sigma'(z) = \sigma(z)(1-\sigma(z))$, где $\sigma(z) = \dfrac{1}{1+e^{-z}}$, и проверьте численно.

`modules/01_notation.md`

```python
import numpy as np

sigma = lambda z: 1 / (1 + np.exp(-z))
```

```python
h = 0.0001
for z in [0, 2]:
    print(z, (sigma(z + h) - sigma(z)) / h, sigma(z) * (1 - sigma(z)))
```

## Саммари

**Что разобрали:**
- **Хвост из 0.3:** $\big((x-3)^2\big)' = 2(x-3) \cdot 1 = 2x - 6$ через цепное правило; сверили раскрытием скобок $x^2 - 6x + 9$.
- **Разминка (BASICS Б7.1 + одна своя):** $(5x^4 - 2x)' = 20x^3 - 2$; $(e^{-x^2})' = -2x\,e^{-x^2}$; $(\ln 3x)' = \frac1x$; $\left(\frac{1}{1+x^2}\right)' = -\frac{2x}{(1+x^2)^2}$.
- **Доказательство 1.1:**

$$\sigma'(z) = \underbrace{-\frac{1}{(1+e^{-z})^2}}_{df/du} \cdot \underbrace{(-e^{-z})}_{du/dz} = \frac{e^{-z}}{(1+e^{-z})^2} = \underbrace{\frac{1}{1+e^{-z}}}_{\sigma(z)} \cdot \underbrace{\frac{e^{-z}}{1+e^{-z}}}_{1-\sigma(z)}$$

  Вторую дробь узнали, вычтя $\sigma$ из единицы: $1 - \frac{1}{1+e^{-z}} = \frac{1+e^{-z}-1}{1+e^{-z}} = \frac{e^{-z}}{1+e^{-z}}$.
- **Численно:** $\frac{\sigma(z+h)-\sigma(z)}{h}$ совпадает с $\sigma(z)(1-\sigma(z))$ с погрешностью, которая падает при уменьшении $h$.
- **ML:** сигмоида превращает логит в вероятность класса 1. Её производная нужна градиентному спуску, и формула $\sigma(1-\sigma)$ дешёвая: $\sigma$ уже посчитана на прямом проходе. Backprop — это цепное правило через все слои сети (модуль 04).

### Теория по трудным местам

**Производная** — во сколько раз быстрее аргумента меняется функция около точки: $\Delta f \approx f'(x)\,\Delta x$. Запись $\frac{df}{dx}$ означает то же, что $f'$.

**Таблица и правила сборки** (`BASICS.md`, раздел Б7):
- $(c)' = 0$; $(x^n)' = n x^{n-1}$; $(e^x)' = e^x$; $(\ln x)' = \frac1x$.
- $(c f)' = c f'$ (множитель остаётся); $(f+g)' = f' + g'$ (сумма по частям).
- $x = x^1$, $a^0 = 1$, поэтому $(x)' = 1$.

**Правило степени — только для числового показателя.** $x^n$: основание переменная, показатель число. В $e^{z}$ наоборот, и для неё своя строка: $(e^z)' = e^z$.

**Минус перед квадратом:** $-x^2 = -(x^2) = -1 \cdot x^2$, производная $-2x$. Это не $(-x)^2 = x^2$.

**Дробь как степень:** $\frac{1}{u} = u^{-1}$, $\left(\frac1u\right)' = -\frac{1}{u^2}$. Степень $-1$ действует на весь знаменатель: $\frac{1}{1+x^2} = (1+x^2)^{-1}$, а не $1 + x^{-2}$.

**Внешняя и внутренняя функции — метод калькулятора.** Посчитай выражение для конкретного числа. Последнее действие — внешняя функция, всё, что до него, — внутренняя $u$.

**Цепное правило:**

$$\frac{df}{dx} = \frac{df}{du} \cdot \frac{du}{dx}$$

- Каждое звено усиливает то, что пришло к нему, поэтому усиления **перемножаются**: $(3x)^2$ при $x=1$: $\Delta x = 0.01 \to \Delta u = 0.03$ (×3) $\to \Delta f = 0.18$ (×6), итого ×18.
- $u$ — временное имя внутренней части. В конце подставляем его обратно; вместо $x$ ничего не подставляем.
- Звеньев может быть сколько угодно, тогда перемножаются производные всех звеньев.

**Знаки:** $(-a)\cdot(-b) = ab$.

**Вычесть дробь из единицы:** $1 = \frac{A}{A}$, поэтому $1 - \frac{1}{A} = \frac{A-1}{A}$. Пример: $1 - \frac14 = \frac34$.

**Квадрат разности:** $(a-b)^2 = a^2 - 2ab + b^2$.

**Python:** деление выполняется раньше вычитания, поэтому разность в скобки: `(f(x+h) - f(x)) / h`.

**Хвосты:** нет.
